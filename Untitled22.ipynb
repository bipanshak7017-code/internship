```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go


sns.set_theme(style="whitegrid")
print("Libraries successfully loaded!")
```

```python

df_nav = pd.read_csv('02_nav_history.csv')
df_aum = pd.read_csv('03_aum_by_fund_house.csv')
df_inflows = pd.read_csv('05_category_inflows.csv')
df_holdings = pd.read_csv('09_portfolio_holdings.csv')

print("Data loaded successfully!")
df_nav.head()
```

```python
df_nav['date'] = pd.to_datetime(df_nav['date'])

fig = px.line(df_nav, x='date', y='nav', title='NAV Trend (2022-2026)')

fig.add_vrect(x0="2023-01-01", x1="2023-12-31",
              annotation_text="2023 Bull Run", fillcolor="green", opacity=0.1, line_width=0)

fig.add_vrect(x0="2024-01-01", x1="2024-12-31",
              annotation_text="2024 Correction", fillcolor="red", opacity=0.1, line_width=0)

fig.show()
```

```python
# Task 4: Category Inflow Heatmap
# Ensure '05_category_inflows.csv' is loaded as 'df_inflows'

# 1. Pivot the data: rows = Category, columns = Month, values = Net_Inflow
pivot_df = df_inflows.pivot_table(index='category', columns='month', values='net_inflow_crore')

# 2. Create the heatmap
plt.figure(figsize=(14, 8))
sns.heatmap(pivot_df, annot=True, fmt='.1f', cmap='YlGnBu')
plt.title('Category-wise Net Inflow Heatmap')
plt.show()
```

```python
# Task 8: NAV Return Correlation Matrix
# 1. Pivot the data to get NAV for each scheme
pivot_nav = df_nav.pivot(index='date', columns='amfi_code', values='nav')

# 2. Calculate Daily Returns
returns = pivot_nav.pct_change()

# 3. Correlation Matrix
corr_matrix = returns.corr()

# 4. Plot Heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=False, cmap='coolwarm')
plt.title('NAV Return Correlation Matrix')
plt.show()
```

```python

sector_data = df_holdings.groupby('sector')['weight_pct'].sum()


plt.figure(figsize=(8, 8))
plt.pie(sector_data, labels=sector_data.index, autopct='%1.1f%%', pctdistance=0.85, wedgeprops={'width': 0.3})
plt.title('Sector Allocation')
plt.show()
```

```python
# Task 2: AUM Growth Bar Chart
plt.figure(figsize=(14, 7))
sns.barplot(data=df_aum, x='fund_house', y='aum_crore', hue='date')
plt.title('AUM Growth by Fund House (2022-2025)')
plt.ylabel('AUM (in Crore)')
plt.xticks(rotation=45)
plt.show()
```

```python
# Task 3: SIP Inflow Time-Series
df_sip = pd.read_csv('04_monthly_sip_inflows.csv')
df_sip['month'] = pd.to_datetime(df_sip['month'])

fig = px.line(df_sip, x='month', y='sip_inflow_crore', title='Monthly SIP Inflow Trend')
fig.add_annotation(x='2025-12-01', y=31002, text="All-time High: ₹31,002 Cr", showarrow=True)
fig.show()
```
